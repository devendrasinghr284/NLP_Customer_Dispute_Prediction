```python
# ====================================================
# STEP 2: Load Groq API key from Colab Secrets
# ====================================================

import os
from google.colab import userdata

groq_key = userdata.get('apikey3')

if not groq_key:
    raise ValueError('apikey3 not found in Colab Secrets')

# Set standard Groq environment variable
os.environ['GROQ_API_KEY'] = groq_key

print('Groq API key loaded successfully')
```

```python
# ====================================================
# STEP 3: Initialize Groq LLM
# ====================================================

from langchain_groq import ChatGroq
from langchain_core.output_parsers import StrOutputParser

llm = ChatGroq(
    model='llama-3.1-8b-instant',
    temperature=0.3
)

parser = StrOutputParser()

chain = llm | parser

print('Groq model initialized successfully')
```

```python
# ====================================================
# STEP 4: Upload multiple PDF files
# ====================================================

from google.colab import files

uploaded = files.upload()

# Keep only PDF files
pdf_files = [f for f in uploaded.keys() if f.lower().endswith('.pdf')]

if not pdf_files:
    raise ValueError('No PDF files were uploaded')

print('Uploaded PDF files:')

for pdf in pdf_files:
    print('-', pdf)
```

```python
# ====================================================
# STEP 5: Read and combine all PDF files
# ====================================================

from pypdf import PdfReader

combined_text = ''

for pdf_path in pdf_files:
    print(f'Reading: {pdf_path}')

    reader = PdfReader(pdf_path)

    for page in reader.pages:
        page_text = page.extract_text()

        if page_text:
            combined_text += page_text + '\n'

print('\nTotal combined characters:', len(combined_text))
```

```python
# ====================================================
# STEP 6: Clean extracted text
# ====================================================

clean_text = ' '.join(combined_text.split())

print('Preview of combined text:\n')
print(clean_text[:3000])
```

```python
# ====================================================
# STEP 7: Generate MCQ quiz from combined PDF content
# ====================================================

# Use a reasonable portion of the text
sample_text = clean_text[:10000]

prompt = f'''
You are an expert instructor in Large Language Models (LLMs), LangChain, and Prompt Engineering.

The following content comes from uploaded PDF documents that discuss:

- LLM Input / Output Workflow
- LLMs and Chat Models
- Prompting with Prompt Templates
- Formatting LLM Responses with Output Parsers
- Other Advanced LLM Operations

Create a high-quality quiz for engineering or data science students.

Instructions:

1. Generate **20 MCQ questions**.
2. Each question must have **4 options** labeled **A, B, C, and D**.
3. Only **one option should be correct**.
4. After each question provide:
   - **Correct Answer**
   - **Short Explanation** (1-2 lines)
5. Questions should test both **conceptual understanding and practical usage**.
6. Focus especially on:
   - LLM workflows
   - Chat models vs traditional LLMs
   - Prompt templates
   - Output parsers
   - Chaining and structured outputs

PDF Content:
{sample_text}
'''

quiz_output = chain.invoke(prompt)

print(quiz_output)
```

```python
# ====================================================
# STEP 8: Save quiz output
# ====================================================

output_file = 'LLM_LangChain_MCQ_Quiz.txt'

with open(output_file, 'w', encoding='utf-8') as f:
    f.write(quiz_output)

print(f'Quiz saved as: {output_file}')
```
